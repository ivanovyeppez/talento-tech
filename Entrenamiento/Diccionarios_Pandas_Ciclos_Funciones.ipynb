## 5) 🧪 Laboratorio Final Integrador (35–40 min)

### Objetivo
Integrar **diccionarios + pandas + ciclos + funciones**.

### Instrucciones
1. Crea una **lista de 6 diccionarios** con `nombre`, `edad`, `nota` (decimal) y `ciudad`.
2. Conviértela a `DataFrame` (`df_lab`).
3. Recorre la tabla e imprime `"Nombre - Ciudad - Nota"`.
4. Crea una función `mejores(df, umbral)` que **retorne** un nuevo DataFrame con notas `>= umbral`.
5. Aplica `mejores(df_lab, 4.0)` y guarda el resultado en `top`.
6. Calcula el **promedio** de `nota` de `top` y muéstralo.
7. Exporta `top` a un archivo CSV llamado `mejores_estudiantes.csv`.

> Extra (opcional): añade una columna `estado` con el texto `"APROBADO"` si `nota >= 4.0`, en caso contrario `"EN PROCESO"`.

```python
# Escribe tu solución del laboratorio aquí
# Paso 1: datos = [ {...}, ... ]
# Paso 2: df_lab = pd.DataFrame(datos)
# Paso 3: recorrer con for / iterrows()
# Paso 4: def mejores(df, umbral): ...
# Paso 5: top = mejores(df_lab, 4.0)
# Paso 6: promedio_top = ...
# Paso 7: exportar CSV -> df.to_csv("mejores_estudiantes.csv", index=False)
# Extra: crear columna 'estado'
# TODO

import pandas as pd

# 1. Lista de 6 diccionarios
estudiantes = [
    {"nombre": "Ana",   "edad": 20, "nota": 4.5, "ciudad": "Pasto"},
    {"nombre": "Luis",  "edad": 22, "nota": 3.8, "ciudad": "Medellín"},
    {"nombre": "Marta", "edad": 21, "nota": 4.2, "ciudad": "Cali"},
    {"nombre": "Juan",  "edad": 23, "nota": 3.9, "ciudad": "Pereira"},
    {"nombre": "Sofia", "edad": 20, "nota": 3.5, "ciudad": "Bogotá"},
    {"nombre": "Pedro", "edad": 24, "nota": 4.0, "ciudad": "Manizales"}
]

# 2. Convertir a DataFrame
df_lab = pd.DataFrame(estudiantes)

# 3. Recorrer la tabla e imprimir "Nombre - Ciudad - Nota"
print("Nombres, Ciudades, Notas:")
for _, fila in df_lab.iterrows():
    print(f"{fila['nombre']} - {fila['ciudad']} - {fila['nota']}")

# 4. Función mejores(df, umbral)
def func_mejores(df, umbral):
    return df[df["nota"] >= umbral]

# 5. Filtrar con umbral = 4.0
print("\n--------------------------------------------")
top = func_mejores(df_lab, 4.0)
print("Type Top= ",type(top))

# 6. Calcular promedio de nota de top
print("\n--------------------------------------------")
promedio_grl = df_lab["nota"].mean()
print("Promedio de nota (general):", promedio_grl)

promedio_top = top["nota"].mean()
print("Promedio de nota (top):", promedio_top)

# 7. Exportar top a CSV
top.to_csv("mejores_estudiantes.csv", index=False)

# Extra: añadir columna estado
df_lab["estado"] = df_lab["nota"].apply(lambda x: "APROBADO" if x >= 4.0 else "EN PROCESO")
print("\nDataFrame con estado:\n", df_lab)
```
